# Seaborn

A visualization library that sits on top of matplotlib.

````python
%matplotlib inline
import pandas as pd

# Load the data from the CSV file
df = pd.read_csv("http://media.sundog-soft.com/SelfDriving/FuelEfficiency.csv")

# Count the number of cars for each number of gears and plot the results
gear_counts = df['# Gears'].value_counts()
gear_counts.plot(kind='bar')
````

#### We can use Seaborn and change the default matplotlib settings, to make this more visually appealing and add more features

````python
import seaborn as sns
sns.set()
````

### Now if we use same plot command, it will be more better looking

````python
gear_counts.plot(kind='bar')
````

## DistPlot (Depriciated)
A special add on over matplotlib. Seaborn offers this kind of plot, which combines a histogram with a distribution curve.

* it is a depriciated function. Use `displot` (a figure-level function with
similar flexibility) or `histplot` (an axes-level function for histograms).

````python
sns.distplot(df['CombMPG'])
````

### *Alternate to distplot -  histplot

histplot works the same as distplot. In order to see curve lines atop bar graph, set ```kde = true```

````python
sns.histplot(df['CombMPG'], kde=True)
````

## PairPlot
It allows you to vizualize plots of every combination of various attributes together, so you can look for patterns between attributes.

````python
df2=df[['Cylinders','CityMPG', 'HwyMPG', 'CombMPG']]
df2.head()
````

````python
sns.pairplot(df2, hue='Cylinders', height=2.5)
````

## ScatterPlot
Plots individual data points across 2 axes, shows distribution of data along those attributes

````python
sns.scatterplot(x="Eng Displ", y="CombMPG", data=df)
````

## Jointplot
It combines scatterplot with bar graphs on both axes, allowing you to see both individual data points and distribution across both dimensions at the same time. Adding ```kind='reg'``` also adds in linear regression to your graph.

````python
sns.jointplot(x="Eng Displ", y="CombMPG", data=df)
# sns.jointplot(x="Eng Displ", y="CombMPG", data=df, kind='reg')
````

## Lmplot
A scatterplot, but with linear regression line computed and overlaid onto the data

````python
# sns.lmplot(x="Eng Displ", y="CombMPG", data=df, hue='Cylinders', height=5, aspect=3)
sns.lmplot(x="Eng Displ", y="CombMPG", data=df)
````

## Boxplot

Box-and-whiskers plot. Used to study about the spread and skew of the data. Same as the Box-and-whiskers plot of matplotlib

````python
# Set the figure size for the boxplot
sns.set_theme(rc={"figure.figsize": (15, 5)})
ax=sns.boxplot(x="Mfr Name", y="CombMPG", data=df)
# Since the number of manufacturers is large, we will rotate the x-axis labels for better readability
ax.set_xticklabels(ax.get_xticklabels(), rotation=45)
````

## Swarmplot
Like boxplot, but points individual data points, and groups them together based on their distribution.

````python
# sns.set_theme(rc={"figure.figsize": (25, 5)})
ax=sns.swarmplot(x="Mfr Name", y="CombMPG", data=df)
ax.set_xticklabels(ax.get_xticklabels(), rotation=90)
````

## Countplot
A histogram, but for categorical data. Tells you how many times each category on the X axis occurs in the data.

````python
ax=sns.countplot(x="Mfr Name", data=df)
ax.set_xticklabels(ax.get_xticklabels(), rotation=45)
````

## Heatmap
Plots 2D data, where colors represent individual values in each cell

````python
# Creating a pivot table to analyze the relationship between Cylinders, Engine Displacement, and Combined MPG
# lets say that there are 20 cars, with 4 cylinder and 2.0l engine displacement.
# We calculate the corresponsing CombMPG* for this combination
# which will be the average of the combMPG values for all 20 cars with this combination.

df2=df.pivot_table(index='Cylinders', columns='Eng Displ', values='CombMPG', aggfunc='mean')
sns.heatmap(df2)
````
