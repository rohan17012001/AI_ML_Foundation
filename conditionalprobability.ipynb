# Conditional Probability

Let's say A and B are two events. Let $P(A)$ and $P(B)$ be the individual probabilities of these events. Let $P(A, B)$ be the probability of both the events occuring together.
The probability $P(B|A)$, which means the probability of event B occuring if the event A has occured is given by:

$ P(B|A) = \frac{P(A,B)}{P(A)} $


For our experiment, we are measuring the probability of people from a certain age group buying a product.

We have the following age groups: 20s, 30s, 40s, 50s, 60s, 70s

We have 2 dictionaries:
- totals, which stores the count of total people from each age group, and
- purchases, which stores the count of people who bought the product from each age group.

```python
from numpy import random
random.seed(0)

totals = {20:0, 30:0, 40:0, 50:0, 60:0, 70:0}
purchases = {20:0, 30:0, 40:0, 50:0, 60:0, 70:0}
totalPurchases = 0
for _ in range(100000):
    ageDecade = random.choice([20, 30, 40, 50, 60, 70])
    
    # We are calculating the probability of each age group buying our stuff.
    # So, a 20 year old will have a probability of 20%, 30 year will have 30%, and so on

    purchaseProbability = float(ageDecade)/100
    totals[ageDecade]+=1

    # random.random() gives a number between 0 and 1
    # If its value is less than purchaseProbability, it means that the person bought the product.
    # eg: if random.random()=0.11 and purchaseProbability = 0.20, it means that the person bought the product
    # and vice versa.

    if(random.random()<purchaseProbability):
        totalPurchases+=1
        purchases[ageDecade]+=1
    
```
