# Linear Regression

We create a sample data to show a roughly linear relationship between page render speed and successful purchases

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from pylab import *

# Pagespeeds centered around 3 seconds, with a standard deviation of 1 second
pageSpeeds = np.random.normal(3.0, 1.0, 1000)
purchaseAmount = 100 - (pageSpeeds + np.random.normal(0, 0.1, 1000))*3
scatter(pageSpeeds, purchaseAmount)
plt.show()
```

Now we can apply linear regression using scipy.lingress

```python
from scipy import stats

slope, intercept, r_value, p_value, std_err = stats.linregress(pageSpeeds, purchaseAmount)
```

And we get a near perfect value for our R-squared value

```python
r_value **2
```

Now we can use the slope and intercept we got from the regression to plot predicted values vs observed

```python
def predict(x):
    return slope * x + intercept

# Calculating the predicted values of purchaseAmount based on the pageSpeeds
fitLine = predict(pageSpeeds)
plt.scatter(pageSpeeds, purchaseAmount)
# Plotting the regression line
plt.plot(pageSpeeds, fitLine, c='r')
plt.show()
```
