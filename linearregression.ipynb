# Linear Regression

We create a sample data to show a roughly linear relationship between page render speed and successful purchases

```python
%matplotlib inline
import numpy as np
import matplotlib.pyplot as plt
from pylab import *

pageSpeeds = np.random.normal(3.0, 1.0, 1000)
purchaseAmount = 100 - (pageSpeeds + np.random.normal(0, 0.1, 1000))*3
scatter(pageSpeeds, purchaseAmount)
plt.show()
```
